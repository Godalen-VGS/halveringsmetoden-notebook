## Ideen

Hvis $f$ er kontinuerlig og $f(a)$ og $f(b)$ har **ulikt fortegn**, må grafen krysse $x$-aksen et sted mellom $a$ og $b$.

1. Finn midtpunktet $m = \dfrac{a + b}{2}$.
2. Hvis $f(a)$ og $f(m)$ har ulikt fortegn, ligger nullpunktet i $[a, m]$. Sett $b = m$.
3. Ellers ligger det i $[m, b]$. Sett $a = m$.
4. Gjenta til intervallet er kortere enn ønsket nøyaktighet.

For hvert steg halveres intervallet. Derav navnet.

Tips: To tall har ulikt fortegn hvis produktet av dem er negativt.

> Behold funksjonsnavnene og parameterne nøyaktig slik oppgaven sier. Det er det som gjør at selvtesten og den automatiske rettingen kan sjekke svarene dine.

## Test løsningen din

````python
import math

def test_halvering():
    tilfeller = [
        (lambda x: x**2 - 2, 0, 2, math.sqrt(2)),
        (lambda x: x**2 - 2, -2, 0, -math.sqrt(2)),
        (lambda x: x**3 - x - 2, 1, 2, 1.5213797068045676),
        (lambda x: math.cos(x) - x, 0, 1, 0.7390851332151607),
        (lambda x: 3*x - 6, 0, 5, 2.0),
    ]
    feil = 0
    for f, a, b, fasit in tilfeller:
        svar = halvering(f, a, b)
        if svar is None or abs(svar - fasit) > 1e-6:
            print(f"❌ Intervall [{a}, {b}]: fikk {svar!r}, forventet ca. {fasit}")
            feil += 1
    if halvering(lambda x: x**2 + 1, -1, 1) is not None:
        print("❌ x² + 1 har ikke noe nullpunkt i [-1, 1], så svaret skal være None")
        feil += 1
    svar = halvering(lambda x: x**2 - 2, 0, 2, toleranse=1e-10)
    if svar is None or abs(svar - math.sqrt(2)) > 1e-10:
        print(f"❌ Med toleranse=1e-10 fikk du {svar!r}, som ikke er nøyaktig nok")
        feil += 1
    if feil == 0:
        print("✅ Alle testene bestått!")

test_halvering()
````
